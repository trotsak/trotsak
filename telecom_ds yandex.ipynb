# Выпускной проект: "Телекоммуникация".

Оператор связи «ТелеДом» хочет бороться с оттоком клиентов. Для этого его сотрудники начнут предлагать промокоды и специальные условия всем, кто планирует отказаться от услуг связи. Чтобы заранее находить таких пользователей, «ТелеДому» нужна модель, которая будет предсказывать, разорвёт ли абонент договор. Команда оператора собрала персональные данные о некоторых клиентах, информацию об их тарифах и услугах. 

### Задача

Обучить на предоставленых данных модель классификации для прогноза оттока клиентов c **AUC-ROC >= 0.85** 

### Описание услуг

Оператор предоставляет два основных типа услуг: 
- Стационарную телефонную связь. Телефон можно подключить к нескольким линиям одновременно.
- Интернет. Подключение может быть двух типов: через телефонную линию (`DSL`, от англ. digital subscriber line — «цифровая абонентская линия») или оптоволоконный кабель (`Fiber optic`).

Также доступны такие услуги:
- Интернет-безопасность: антивирус (`DeviceProtection`) и блокировка небезопасных сайтов (`OnlineSecurity`);
- Выделенная линия технической поддержки (`TechSupport`);
- Облачное хранилище файлов для резервного копирования данных (`OnlineBackup`);
- Стриминговое телевидение (`StreamingTV`) и каталог фильмов (`StreamingMovies`).

Клиенты могут платить за услуги каждый месяц или заключить договор на 1–2 года. Возможно оплатить счёт разными способами, а также получить электронный чек.

### План работы

**1. Загрузка и подготовка данных**

- Загрузим датасет, изучим его.
- Посмотрим на пропуски, аномалии, оформление.

**2. Предобработка данных**
- Оформим по стандартам.
- Заполним пропуски.
- Исправим типы данных.
- Добавим целевой прзнак.

**3. Исследовательский анализ**

- Посмотрим соотношения, доли признаков.
- Проанализируем каждый датасет отдельно.

**4. Корреляционный анализ**

- Выявим взаимосвязи признаков.
- Обозначим полезные для модели признаки. 
- Удалим малозначимые признаки.

**6. Объеденение данных**

- Объеденим данные в один датасет.


**6. Создание новых признаков**

- Создадим новые признаки.

**7. Исследовательский и корреляционный анализ объединенного датасета**

- Исследуем новый единый датасет, распределение признаков.
- Посмторим как коррелируют признаки между собой.
- Удалим мультиколлинеарность.
- Оставим только необходимые признаки для модели.


**4. Разделение данных**

- Разделим 75% данных на обучение, 25% — для тестирования.

**5. Обучение моделей**

### Изучим общую информацию о полученных датафреймах
Создадим функцию `data_info` для вывода общей информации по датафрейму.

```python
# создание функции для вывода общей информации по датафрейму
def data_info(data, dataframe_name):
    """
    Отображает общую информацию о переданном датафрейме.

    Функция выполняет следующие операции:
    1. Отображение первых нескольких строк датафрейма.
    2. Вывод общей информации о датафрейме, включая типы данных и количество ненулевых значений.
    3. Отображение статистического описания числовых столбцов.
    4. Подсчет и вывод количества пропущенных значений в каждом столбце.
    5. Вывод количества явных дубликатов в датафрейме.
    6. Отображение списка названий столбцов в датафрейме.
    7. Вывод уникальных значений для столбцов с типом данных 'object'.
    8. Вывод числа уникальных значений для каждого столбца.
    9. Вывод числа дублей для каждого столбца.

    Параметры:
    ----------
    data : pandas.DataFrame
        Датафрейм, для которого необходимо вывести информацию.
    dataframe_name : str
        Имя датафрейма (для отображения в выводе).
    """

    # отображение первых несколько строк датафрейма
    print(Color.BOLD + f"Первые строки датафрейма {dataframe_name}:\n" + Color.END)
    display(data.head())
    print()

    # вывод информацию о датафрейме, включая типы данных и количество ненулевых значений
    print(Color.BOLD + f"Общая информация о датафрейме {dataframe_name}:\n" + Color.END)
    data.info()
    print()

    # отображение статистического описания числовых столбцов датафрейма
    print(Color.BOLD + f"Статистическое описание числовых столбцов датафрейма {dataframe_name}:\n" + Color.END)
    display(data.describe())
    print()

    # отображение количества пропущенных значений в каждом столбце
    print(Color.BOLD + f"Количества пропущенных значений в каждом столбце датафрейма {dataframe_name}:\n" + Color.END)
    display(data.isna().sum())
    print()

    # вывод количества явных дубликатов в датафрейме
    print(f'Количество явных дубликатов в датафрейме: {Color.RED}{data.duplicated().sum()}{Color.END}.')
    print()

    # отображение списка названий столбцов в датафрейме
    print(Color.BOLD + f"Cписок названий столбцов в датафрейме {dataframe_name}:\n" + Color.END)
    display(data.columns.tolist())

    # отображение всех уникальных значений и их количества в столбцах типа 'object'
    for i in data.columns:
        if data[i].dtype == 'object':
            unique_values = data[i].unique()
            num_unique_values = len(unique_values)
            num_duplicates = data[i].duplicated().sum()
            print(f'В столбце {Color.BOLD}\'{i}\'{Color.END} содержится {num_unique_values} уникальных значений: \
            {Color.BOLD}{unique_values}{Color.END}')
            print(f'Число дублей в столбце {Color.BOLD}\'{i}\'{Color.END}: {Color.RED}{num_duplicates}{Color.END}')
```

### Статистический анализ всех признаков.
Создадим функции для построения графиков количественных и категориальных признаков.

```python
# создание функции для построения графиков количественных признаков
def graphs_col(df, feature, col_name_ru, number_bins):
    
    plt.figure(figsize=(10, 8))
    
    # 1. построение графика распределения
    plt.subplot(2, 1, 1)
    # построение гистограммы
    counts, bins, _ = plt.hist(df[feature], edgecolor='white', bins=number_bins, density=True)
    
    
    # вычисление параметров нормального распределения
    mu = np.mean(df[feature])  # среднее
    sigma = np.std(df[feature])  # стандартное отклонение
    
    # создание диапазона значений для нормальной кривой
    x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
    # нормальная кривая
    plt.plot(x, stats.norm.pdf(x, mu, sigma), color='red', linewidth=2, label='Нормальная кривая')
    
    plt.title(f'Распределение признака {feature}', fontsize=15, pad=20)
    plt.ylabel('Плотность распределения', fontsize=14)
    plt.xlabel(f'{col_name_ru}', fontsize=14)
    plt.legend()    
   
    # 2. построение диаграммы размаха
    plt.subplot(2, 1, 2)
    df.boxplot(column=feature, fontsize=14, vert=False)
    plt.title(f'Диаграмма размаха для параметра {feature}', fontsize=15, pad=20)
    plt.xlabel('')
    plt.ylabel(f'{col_name_ru}', fontsize=14)
    
    plt.yticks([])


     
    plt.tight_layout(pad=1.0)

    plt.subplots_adjust(hspace=0.3)
    plt.show()
    
    # вычисление пика и центра бина
    max_value = np.max(counts)  # максимальное значение частоты
    peak_bin_index = np.argmax(counts)  # индекс максимальной частоты
    peak_bin_center = (bins[peak_bin_index] + bins[peak_bin_index + 1]) / 2  # Центр пика
    
    print(f'Максимальная частота: {max_value:.2f}')
    print(f'Пик (центр бина): {peak_bin_center:.2f}\n')
    
    
    # вычисление квантильных значений
    Q1 = df[feature].quantile(0.25)  # нижний квартиль
    Q3 = df[feature].quantile(0.75)  # верхний квартиль
    IQR = Q3 - Q1                    # интерквартильный размах
    
    
    # вычисление значений усов для диаграммы размаха (минимальные и максимальные значения в пределах 1.5 IQR)
    lower_whisker = df[feature][(df[feature] >= Q1 - 1.5 * IQR)].min()
    upper_whisker = df[feature][(df[feature] <= Q3 + 1.5 * IQR)].max()   
    
    print(f'Нижний ус (whisker): {lower_whisker}')
    print(f'Верхний ус (whisker): {upper_whisker}\n')
   
   
   
 # проверка на нормальность распределения
    if len(df[feature]) < 5000:
        # тест Шапиро-Уилка
        stat_shapiro, p_value_shapiro = stats.shapiro(df[feature])
        alpha = 0.05  # уровень значимости
        print('Тест Шапиро-Уилка:')
        print('Статистика:', stat_shapiro)
        print('p-значение:', p_value_shapiro)
        if p_value_shapiro > alpha:
            print("Вероятно распределение нормально (не отклоняем H0)")
        else:
            print("Вероятно распределение ненормально (отклоняем H0)")
    else:    
            # тест Андерсона-Дарлинга
        result = stats.anderson(df[feature])
        print('\nТест Андерсона-Дарлинга:')
        print('Статистика:', result.statistic)
        for i in range(len(result.critical_values)):
            print(f'Критическое значение при уровне значимости {result.significance_level[i]}: {result.critical_values[i]}')
        if result.statistic < result.critical_values[2]:  # 5% уровень значимости
            print("Вероятно распределение нормально (не отклоняем H0)")
        else:
            print("Вероятно распределение ненормально (отклоняем H0)")
        
    # поиск выбросов
    # пределы для выбросов
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers_indices = df[(df[feature] < lower_bound) | (df[feature] > upper_bound)].index.tolist()
    print(f'\nВсего строк с выбросами для {feature}: {len(outliers_indices)}')
    print(f'Строки с выбросами составляют от всего датафрейма: {round((len(outliers_indices) / len(df))*100, 2)} %\n')
```

```python
def graphs_cat(df, feature, col_name_ru):
    """
    Визуализирует категориальный признак (с числом уникальных значений < 4):
    - Гистограмма с аннотациями
    - Круговая диаграмма с подписями процентов

    Параметры:
        df : pd.DataFrame — исходный датафрейм
        feature : str — имя категориального признака
        col_name_ru : str — название признака (на русском) для заголовков

    Возвращает:
        None — только отображает графики
    """
    counts = df[feature].value_counts()
    plt.figure(figsize=(10, 5))
    # Увеличиваем общее пространство между элементами
    plt.subplots_adjust(hspace=0.4)

    # Столбчатая диаграмма
    plt.subplot(1, 2, 1)
    bars = plt.bar(counts.index.astype(str), counts.values, color='#076fa2')
    plt.title(f'{col_name_ru}, количество', fontsize=15, pad=20)
    plt.xticks(rotation=0, fontsize=13)
    plt.yticks(fontsize=12)
    plt.xlabel(f'{col_name_ru}', fontsize=14)
    plt.ylabel('Количество', fontsize=14)

    # Аннотации
    for bar in bars:
        height = bar.get_height()
        plt.annotate(f'{int(height)}',
                     xy=(bar.get_x() + bar.get_width() / 2, height),
                     xytext=(0, 3),  # смещение текста вверх
                     textcoords="offset points",
                     ha='center', va='bottom', fontsize=12, color='black', fontweight='bold')

    # Круговая диаграмма
    plt.subplot(1, 2, 2)
    plt.pie(counts.values, labels=counts.index.astype(str),
            autopct=lambda pct: f'{pct:.1f}%',
            colors=['#76c7c0', '#66b3ff', '#99ff99'],
            textprops={'fontsize': 14, 'color': 'white', 'fontweight': 'bold'})
    # Добавляем легенду
    plt.legend(
        loc='upper center',
        bbox_to_anchor=(0.5, -0.15),
#         loc='upper left', 
#         bbox_to_anchor=(1, 1),
        fontsize=12, 
        title="Категории", 
        title_fontsize=14 
    )
    plt.title(f'{col_name_ru}, доли', fontsize=15)
    plt.tight_layout()
    plt.show()
```

```python
# создание функции для построения графиков с числом категориальных признаков более 3
def graphs_cat_5max(df, feature, col_name_ru):
    # Проверка на количество уникальных значений
    if df[feature].nunique() <= 3:
        print("Число уникальных значений меньше или равно 3. Пожалуйста, используйте функцию graphs_cat.")
        return
    
    plt.figure(figsize=(10, 5))
    
    # 1. Построение столбчатого графика
    plt.subplot(2, 1, 1)
    df.groupby(feature)[feature].count().plot(kind='bar')
    plt.title(f'{col_name_ru} / количество', fontsize=16, pad=20)
    plt.xticks(rotation=55, fontsize=13)
    plt.xlabel(f'{col_name_ru}', fontsize=15)
    plt.ylabel('Количество', fontsize=15)
    
    ax = plt.gca()
    
    # добавление аннотаций значений над столбцами
    for p in ax.patches:
        height = p.get_height()
        ax.annotate(
            f'{int(height)}',  # текст: высота столбца
            (p.get_x() + p.get_width() / 2., height),  # позиция текста
            ha='center', va='bottom', fontsize=12, color='black', fontweight='bold'
        )
    
    # 2. Получение данных для горизонтального графика
    counts = df.groupby(feature)[feature].count().sort_values(ascending=True)
    names = counts.index
    total = counts.sum()
    

    # определение мест для столбиков
    y = [i * 0.9 for i in range(len(names))]

    # присваивание переменной для цвета
    BLUE = "#076fa2"

    fig, ax = plt.subplots(figsize=(10, 5))

    ax.barh(y, counts, height=0.55, align="edge", color=BLUE)

    ax.xaxis.set_tick_params(labelbottom=False, labeltop=False, length=0)
    ax.set_xlim((0, counts.max() + 2))  
    ax.set_ylim((0, len(names) * 0.9 - 0.2))

    # определение видимости границ
    ax.spines["right"].set_visible(False)
    ax.spines["top"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_lw(1.5)

    # скрытие подписей оси Y
    ax.yaxis.set_visible(False)

    # добавление названий столбцов
    PAD = 0.3
    for name, count, y_pos in zip(names, counts, y):
        x = 0
        color = "white"
        path_effects = None
        if count < 8:
            x = count
            color = BLUE    
            path_effects = [withStroke(linewidth=6, foreground="white")]
        
        ax.text(
            x + PAD, y_pos + 0.5 / 2, name, 
            color=color, fontsize=16, va="center",
            path_effects=path_effects
        ) 

    # настройка свободного места
    fig.subplots_adjust(left=0.005, right=1, top=0.8, bottom=0.1)

    # добавление названия графика
    fig.text(
        0, 1.025, f"Частота встречаемости признака {feature}, доли", 
        fontsize=16, fontweight="bold"
    )

    # добавление подписей столбиков
    for name, count, y_pos in zip(names, counts, y):
        percentage = (count / total) * 100  
        ax.text(
            count + PAD, y_pos + 0.5 / 2, f"{percentage:.1f}%",  
            color="black", fontsize=16, va="center",
            path_effects=[withStroke(linewidth=1, foreground="grey")]
        )
    plt.tight_layout()
    plt.show()
```

```python
# создание декоратора для функции построения графиков с дискретными значениями исключающий дату '2020-02-01' в столбце 'end_date'
def exclude_date_for_end_date(date_to_exclude='2020-02-01'):
    """
    Декоратор для исключения конкретной даты только при анализе столбца 'end_date'
    
    Параметры:
        date_to_exclude (str): Дата для исключения (по умолчанию '2020-02-01')
    """
    def decorator(func):
        @wraps(func)  # Сохраняем имя и docstring оригинальной функции
        def wrapper(data, feature, *args, **kwargs):
            # Применяем фильтрацию только для столбца 'end_date'
            if feature == 'end_date':
                filtered_data = data[data[feature] != date_to_exclude].copy()
                if filtered_data.empty:
                    print(f"Нет данных после исключения {date_to_exclude}")
                    return None
                return func(filtered_data, feature, *args, **kwargs)
            # Для других столбцов - обычное поведение
            return func(data, feature, *args, **kwargs)
        return wrapper
    return decorator

@exclude_date_for_end_date('2020-02-01')
def graph_discrete(data, feature, col_name_ru, height_num):
    # Проверяем, является ли столбец датой
    if np.issubdtype(data[feature].dtype, np.datetime64):
        # Преобразуем к строке для красивых подписей
        ordered_dates = sorted(data[feature].dropna().unique())
        ordered_dates_str = [d.strftime('%Y-%m-%d') for d in ordered_dates]
        data = data.copy()
        data[feature] = data[feature].dt.strftime('%Y-%m-%d')
        order = ordered_dates_str
    else:
        order = None

    plt.figure(figsize=(10, 5))
    ax = sns.countplot(data=data, x=feature, color='blue', order=order)
    plt.title(f'{col_name_ru} / количество', fontsize=16, pad=20)
    plt.xlabel(col_name_ru, fontsize=15)
    plt.ylabel('Количество', fontsize=15)
    plt.xticks(rotation=45, ha='right')
    ax.set_xticks(ax.get_xticks()[::2])

    total = len(data)
    peak_values = []

    # Получаем категории из данных в нужном порядке
    if order is not None:
        categories = order
    else:
        categories = [str(cat) for cat in data[feature].value_counts().sort_index().index.tolist()]

    for i, p in enumerate(ax.patches):
        height = p.get_height()
        x_position = p.get_x() + p.get_width() / 2.
        label = categories[i] if i < len(categories) else f'cat_{i}'

        if height > 1000 or height > height_num:
            peak_values.append((label, height, (height / total) * 100))

        if height > height_num:
            ax.annotate(f'{height:.0f}',
                        (x_position, height),
                        ha='center', va='bottom', color='black', fontweight='bold',
                        fontsize=10, xytext=(0, 5), textcoords='offset points',
                        bbox=dict(facecolor='white', edgecolor='blue', boxstyle='round,pad=0.3'))

            percentage = (height / total) * 100
            ax.annotate(f'({percentage:.1f}%)',
                        (x_position, height),
                        ha='center', va='bottom', xytext=(0, -20),
                        textcoords='offset points', fontsize=10, color='black', fontweight='bold',
                        bbox=dict(facecolor='white', edgecolor='blue', boxstyle='round,pad=0.3'))

    plt.tight_layout()
    plt.show()

    # Текстовая часть после графика
    value_counts = data[feature].value_counts().sort_values(ascending=False)
    print("\nПиковые значения по признаку:")
    print(f"  {'Категория':<30} | {'Кол-во':>10} | {'% от общего':>12}")
    print("-" * 60)
    for label, count in value_counts.items():
        if count > height_num:
            percent = (count / total) * 100
            print(f"{str(label):<30} | {count:>10} | {percent:>11.1f}%")
```

## Анализ распределения имеющихся и новых признаков в привязке к оттоку
Проанализируем соотношение ушедших и оставшихся клиентов по разным параметрам.  

Для оптимизации работы создадим несколько функций:

```python
# создание функции для визуализации распределения признаков с разделением данных по факту оттока 
def hist_graph(df, list_pl, list_pl_name, suptitle='', p_hue=None):
    """
        Строит матрицу гистограмм для визуализации распределения признаков.

        Функция создает сетку гистограмм для анализа распределения указанных признаков
        с возможностью разделения данных по категориям (например, по факту оттока).

        Параметры:
            df (pd.DataFrame): Исходный датафрейм с данными
            list_pl (list): Список названий столбцов для анализа
            list_pl_name (list): Список понятных названий для подписей графиков
            suptitle (str, optional): Общий заголовок для всех графиков. По умолчанию ''.
            p_hue (str, optional): Название столбца для разделения по цвету. По умолчанию None.

        Возвращает:
            None: Функция только отображает графики

        Примеры использования:
            >>> hist_graph(
                    df=df_clients,
                    list_pl=['age', 'balance', 'duration'],
                    list_pl_name=['Возраст', 'Баланс', 'Длительность'],
                    suptitle='Распределение характеристик клиентов',
                    p_hue='churn'
                )    
        """
    # Установка палитры
    sns.set_palette("pastel") 

    # Определяем количество столбцов и строк
    max_col = 2 
    max_row = (len(list_pl) + max_col - 1) // max_col  

    # Создаем фигуру
    plt.figure(figsize=[11, max_row * 4])
    plt.suptitle(suptitle, ha='center', fontsize='25')
    plt.subplots_adjust(top=0.93, hspace=0.7)

    # Рисуем графики
    for i, param, name in zip(range(len(list_pl)), list_pl, list_pl_name):
        plt.subplot(max_row, max_col, i + 1) 
        fig = sns.histplot(data=df, x=param, hue=p_hue, multiple='stack')
        
        if p_hue:
            plt.legend(title="Отток", labels=["Да", "Нет"])

        # plt.xlabel('Значение', fontsize="15")
        plt.xlabel('', fontsize="15")
        plt.ylabel('Количество', fontsize="15")
        plt.title(name, fontsize="20")

        
        plt.xticks(rotation=30, fontsize=10)
        # plt.yticks(fontsize=10)

    
    plt.show()
```

```python
# создание функции для визуализации доли ушедших клиентов
def graph_portret(df, list_pl, list_pl_name, suptitle='', p_hue=None, max_col=2):
    
    """
    Визуализирует долю ушедших клиентов по различным категориям
    
    Параметры:
        df : pd.DataFrame - исходный датафрейм
        list_pl : list - список столбцов для анализа
        list_pl_name : list - названия для подграфиков
        suptitle : str - общий заголовок
        p_hue : str - столбец с целевой переменной (отток)
        max_col : int - количество колонок в сетке графиков
    """ 
    
    
    max_row = math.ceil(len(list_pl) / max_col)


    # конфигурация общей фигуры
    plt.figure(figsize=(max_col * 5.5, max_row * 5))
    plt.suptitle(suptitle, ha='center', fontsize=30)
    plt.subplots_adjust(top=0.95, hspace=0.6)    
    

    # построение графиков
    for idx, col in enumerate(list_pl):
        plt.subplot(max_row, max_col, idx + 1)

        # вычисление и подготовка данных
        pt_data = df.pivot_table(
            index=col,
            observed=False,
            values=p_hue,
            aggfunc=lambda x: (x.sum() / x.count()) * 100
        )
        df_graph = pd.DataFrame(pt_data).reset_index()

        # используем цветовую палитру для разных цветов столбиков
        palette = sns.color_palette("husl", n_colors=len(df_graph))

        # построение графика
        barplot = sns.barplot(
            data=df_graph,
            x=col,
            y=p_hue,
            hue=col,
            palette=palette,
            dodge=False,
            legend=False
        )

        # добавление подписей процентов на столбики
        for p in barplot.patches:
            barplot.annotate(
                f'{p.get_height():.1f}%',  
                (p.get_x() + p.get_width() / 2, p.get_height()), 
                ha='center',  
                va='bottom',  
                fontsize=12,  
                color='black' 
            )

        # оформление графика
        plt.xlabel('')
        plt.ylabel('Процент', fontsize=15)
        plt.title(list_pl_name[idx], fontsize=16)
        plt.xticks(rotation=30, fontsize=10)

    plt.show()
```

```python
# создание функции для табличного представления распределения признакови с разделением данных по факту оттока 
def create_table(df, param, p_hue):
    # Вычисляем доли ушедших клиентов для текущего признака
    churn_data = (
        df.groupby(param,observed=False)[p_hue]
        .value_counts(normalize=True)
        .mul(100)  # Переводим в проценты
        .reset_index(name='percentage')
    )

    # Форматируем данные для таблицы
    table_data = churn_data.pivot(index=param, columns=p_hue, values='percentage').fillna(0)
    table_data = table_data.round(1)  # Округляем до одного знака после запятой

    # Переименовываем столбцы
    table_data.columns = ["Остались (%)", "Ушли (%)"]

    return table_data
```

### Корреляционный анализ

Для лучшего понимания наших данных, проверки утечки целевого признака проведём корреляционный анализ.

Для оптимизации кода создадим необходимые функции:

```python
# создание функции визуализация матрицы корреляции Phi_K
def graph_matrix_phik(corr_matrix):
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    plt.figure(figsize=(13, 11)) 
    sns.heatmap(
        corr_matrix, 
        annot=True, 
        fmt=".2f",
        cmap='YlGnBu', 
        vmin=-1, 
        vmax=1,
        linewidths=0.5, 
        linecolor='white',
        cbar_kws={"shrink": .8},
        mask=mask
    )
    plt.title('Матрица корреляции Phi_K', fontsize=14)
    plt.xticks(rotation=90, ha='right') 
    plt.yticks(rotation=0)  
    plt.tight_layout()  
    plt.show()
```

```python
# создание функции для поиска пар с сильной корреляцией 
def find_strong_correlations(corr_matrix):
    """
    Находит пары признаков с сильной корреляцией:
    - Полная (1.0 или -1.0)
    - Весьма высокая (0.9-0.99)
    - Высокая (0.7-0.9)
    """
    results = {
        'Полная': [],
        'Весьма высокая': [],
        'Высокая': []
    }
    
    seen_pairs = set()
    
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            feature1 = corr_matrix.columns[i]
            feature2 = corr_matrix.columns[j]
            corr_value = corr_matrix.iloc[i, j]
            abs_corr = abs(corr_value)
            
            pair_key = tuple(sorted((feature1, feature2)))
            if pair_key in seen_pairs:
                continue
            seen_pairs.add(pair_key)
            
            if abs_corr == 1.0:
                results['Полная'].append((feature1, feature2, corr_value))
            elif abs_corr >= 0.9:
                results['Весьма высокая'].append((feature1, feature2, corr_value))
            elif abs_corr >= 0.7:
                results['Высокая'].append((feature1, feature2, corr_value))
    
    for level in results:
        results[level].sort(key=lambda x: abs(x[2]), reverse=True)
    
    print("\nСИЛЬНЫЕ КОРРЕЛЯЦИИ:")
    for level in ['Полная', 'Весьма высокая', 'Высокая']:
        if results[level]:
            print(f"\n{level.upper()}:")
            for feature1, feature2, corr_value in results[level]:
                print(f"  {feature1} <-> {feature2}: {corr_value:.2f}")   
```

```python
# создание функции для поиска пар со слабой корреляцией 
def find_weak_correlations(corr_matrix):
    """
    Находит пары признаков со слабой корреляцией:
    - Слабая (0.1-0.3)
    - Практически отсутствует (<0.1)
    """
    results = {
        'Слабая': [],
        'Практически отсутствует': []
    }
    
    seen_pairs = set()
    
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            feature1 = corr_matrix.columns[i]
            feature2 = corr_matrix.columns[j]
            corr_value = corr_matrix.iloc[i, j]
            abs_corr = abs(corr_value)
            
            pair_key = tuple(sorted((feature1, feature2)))
            if pair_key in seen_pairs:
                continue
            seen_pairs.add(pair_key)
            
            if 0.1 <= abs_corr < 0.3:  # Слабая корреляция
                results['Слабая'].append((feature1, feature2, corr_value))
            elif abs_corr < 0.1:  # Практически отсутствует
                results['Практически отсутствует'].append((feature1, feature2, corr_value))           
           
    
    for level in results:
        results[level].sort(key=lambda x: abs(x[2]), reverse=True)
    
    print("\nСЛАБЫЕ КОРРЕЛЯЦИИ:")
    print(f"\nСлабая:")
    for feature1, feature2, corr_value in results['Слабая']:
        print(f"  {feature1} <-> {feature2}: {corr_value:.2f}")
    
    print(f"\nПрактически отсутствует:") 
    for feature1, feature2, corr_value in results['Практически отсутствует']:
          print(f"  {feature1} <-> {feature2}: {corr_value:.2f}")   
```

```python
# создание функции для автоматического поиска подозрительных признаков
def detect_target_leakage(df, target_column,interval_cols):
    """
    Обнаруживает потенциальную утечку целевой переменной через анализ корреляций.

    Функция вычисляет матрицу корреляций Фика между всеми признаками и целевой переменной,
    затем идентифицирует признаки с подозрительно высокой корреляцией (> 0.95), которые
    могут указывать на утечку информации о целевой переменной.

    Параметры:
        df (pd.DataFrame): Исходный датафрейм с данными
        target_column (str): Название целевой колонки (таргета)
        interval_cols (list): Список числовых колонок для анализа

    Возвращает:
        None: Функция только выводит результаты анализа

    Примеры:
        >>> detect_target_leakage(
                df=customer_data,
                target_column='churn',
                interval_cols=['age', 'balance', 'duration']
            )
        Корреляции с целевой переменной:
        churn         1.000000
        duration      0.876543
        balance       0.234567
        Name: churn, dtype: float64
        Подозрение на утечку: ['duration']
    """
    correlation = df.phik_matrix(interval_cols=interval_cols)[target_column].sort_values(ascending=False)
    print("Корреляции с целевой переменной:")
    print(correlation)

    high_corr_features = correlation[correlation > 0.95].index.tolist()
    high_corr_features.remove(target_column)

    if high_corr_features:
        print(f"Подозрение на утечку: {high_corr_features}")
    else:
        print("Явных утечек целевого признака не обнаружено.")
```

```python
# создание функции для вычисление VIF
def vif_calc(target, df):

    """
    Рассчитать фактор инфляции дисперсии (Variance Inflation Factor, VIF) для признаков в заданном датафрейме.

    Фактор инфляции дисперсии (VIF) используется для оценки мультиколлинеарности между независимыми переменными
    в модели регрессии. Высокие значения VIF указывают на сильную корреляцию одного признака с другими.

    Параметры:
    ----------
    target : str
        Имя целевой переменной (столбца), который будет исключён из расчётов.
    df : pandas.DataFrame
        Датафрейм, содержащий данные для анализа. Должен включать только числовые признаки
        или содержать смешанные типы данных, которые будут обработаны.

    Возвращает:
    ----------
    pandas.DataFrame
        Датафрейм с VIF и именами признаков. Датафрейм будет содержать два столбца:
        - `features`: Имена признаков
        - `VIF`: Значения VIF для каждого признака

    Предупреждения:
    --------------
    - Если в датафрейме есть признаки с бесконечными или NaN значениями,
      они будут удалены из расчётов.
    - Если датафрейм пуст или не содержит числовых признаков после обработки, 
      будет выведено сообщение об ошибке, и функция завершит выполнение.

    Пример:
    -------
    >>> data = pd.DataFrame({
    >>>     'A': [1, 2, 3, 4, 5],
    >>>     'B': [5, 4, 3, 2, 1],
    >>>     'C': [2, 3, 4, 5, 6]
    >>> })
    >>> vif_calc(target='C', df=data)
    Предупреждение: В данных обнаружены NaN или бесконечные значения. Они будут удалены.
    VIF для признаков:
       features  VIF
    0        A  <значение>
    1        B  <значение>

    """

# удаление целевой переменной
    features = df.drop([target], axis=1) if target in df.columns else df

    # убедимся, что в данных остались только числовые значения
    features = features.select_dtypes(include=[np.number])

    # проверка на наличие NaN, бесконечных значений и их обрабатка
    if not np.isfinite(features.values).all():
        print(f"Предупреждение: В данных обнаружены NaN или бесконечные значения. Они будут удалены.")
        features = features.replace([np.inf, -np.inf], np.nan).dropna()

    # проверка, пуст ли DataFrame или содержит ли хотя бы одну колонку
    if features.empty or features.shape[1] == 0:
        print("Ошибка: Входные данные не содержат признаков для расчёта VIF.")
        return

    # создание таблицы для хранения результатов VIF
    vif_data = pd.DataFrame()
    vif_data['features'] = features.columns

    try:
        vif_data['VIF'] = [variance_inflation_factor(features.values, i) for i in range(features.shape[1])]
    except ValueError as e:
        print("Ошибка при расчёте VIF:", e)
        return

    print("\nVIF для признаков:")
    print(vif_data)
```

### Препроцессор

```python
# Создаем препроцессор для числовых и категориальных признаков
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])


# Категориальный трансформер для логистической регрессии (с drop='first')
categorical_transformer_lr = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False, drop='first'))
])

# Категориальный трансформер для XGB (без drop='first'), т.к. деревья нечувствительны к линейной зависимости фичей
# и сохраняются все признаки для анализа их важности
categorical_transformer_xgb = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# categorical_transformer = Pipeline(steps=[
#     ('imputer', SimpleImputer(strategy='most_frequent')),
#     ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
# ])


# Препроцессор для Logistic Regression
preprocessor_lr = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_cols),
        ('cat', categorical_transformer_lr, categorical_cols)
    ],
    remainder='passthrough'
)

# Препроцессор для XGBoost
preprocessor_xgb = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_cols),
        ('cat', categorical_transformer_xgb, categorical_cols)
    ],
    remainder='passthrough'
)


# preprocessor = ColumnTransformer(
#     transformers=[
#         ('num', numeric_transformer, numerical_cols),
#         ('cat', categorical_transformer, categorical_cols)
#     ],
#     remainder='passthrough'
# )
```

## Построение и обучение моделей

### Логистическая регрессия

```python
# паплайн
log_reg_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_lr),
    ('classifier', LogisticRegression(max_iter=5000, solver='saga', random_state=RANDOM_STATE))
])

# сетка параметров
log_reg_param_grid = {
    'classifier__C': np.logspace(-2, 2, 10),  
    'classifier__penalty': ['l1','l2'],    
    'classifier__class_weight': [None, 'balanced', {0: 1, 1: 2}, {0: 1, 1: 5}]
}

# создние объекта StratifiedKFold для кросс-валидации
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

# создание объекта RandomizedSearchCV для поиска оптимальных гиперпараметров модели
log_reg_random = RandomizedSearchCV(
    log_reg_pipeline,
    log_reg_param_grid,
    n_iter=30,
    cv=cv,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1,
    random_state=RANDOM_STATE
)
```

```python
%%time
# обучение модели
log_reg_random.fit(X_train, y_train)
```

```python
# Лучшие параметры и результаты
print(f"{Color.BOLD}\nЛогистическая регрессия:{Color.END}")
print(f"Лучшие параметры: {log_reg_random.best_params_}")
print(f"Лучший {Color.BOLD}ROC-AUC{Color.END} на кросс-валидации: \
{Color.BOLD}{log_reg_random.best_score_:.4f}{Color.END}")
```

**Вывод:** ROC-AUC на кросс-валидации (0.75) не достиг целевого показателя, попробуем другую модель.

### XGB

```python
# XGBoost классификатор
xgb_model = XGBClassifier(
    objective='binary:logistic',
    random_state=RANDOM_STATE,
    eval_metric='logloss',
    tree_method='hist'
)

# паплайн
xgb_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_xgb),
    ('classifier', xgb_model)
])


# расчет веса для положительного класса
scale_pos_weight = y_train.value_counts()[0] / y_train.value_counts()[1]

# сетка параметров
xgb_param_grid = {
    'classifier__n_estimators': [100, 200, 300, 500],
    'classifier__max_depth': [3, 5, 7, 9],
    'classifier__learning_rate': [0.005, 0.01, 0.05, 0.1, 0.2],
    'classifier__subsample': [0.6, 0.8, 1.0],
    'classifier__colsample_bytree': [0.6, 0.8, 1.0],
    'classifier__gamma': [0, 0.1, 0.5, 1],
    'classifier__reg_alpha': [0, 0.1, 1],
    'classifier__reg_lambda': [0, 0.1, 1],
    'classifier__min_child_weight': [1, 3, 5, 7],
    'classifier__scale_pos_weight': [scale_pos_weight, scale_pos_weight*0.5, scale_pos_weight*2]
}

# создние объекта StratifiedKFold для кросс-валидации
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

# создание объекта RandomizedSearchCV для поиска оптимальных гиперпараметров модели
xgb_random = RandomizedSearchCV(
    estimator=xgb_pipeline,
    param_distributions=xgb_param_grid,
    n_iter=50,
    scoring='roc_auc',
    cv=cv,
    n_jobs=-1,
    verbose=1,
    random_state=RANDOM_STATE,
    error_score='raise'
)
```

```python
%%time
# Обучение модели
xgb_random.fit(X_train, y_train)
```

```python
# Лучшие параметры и результаты
print(f"{Color.BOLD}\nXGBoost:{Color.END}")
print(f"Лучшие параметры: {xgb_random.best_params_}")
print(f"Лучший {Color.BOLD}ROC-AUC{Color.END} на кросс-валидации: {Color.BOLD}{xgb_random.best_score_:.4f}{Color.END}")
```

**Вывод:** ROC-AUC на кросс-валидации (0.89) модели XGBoost достиг целевого показателя. Можно тестировать.

## Сравнение моделей

Сравним метрики моделей для выбора лучшей. 

```python
# Вычисляем ROC-AUC на валидационной выборке для каждой модели
log_reg_val_auc = log_reg_random.best_score_
xgb_val_auc = xgb_random.best_score_

# Создаем таблицу с результатами
models_results = pd.DataFrame({
    'Модель': ['Логистическая регрессия', 'XGBoost'],
    'ROC-AUC (валидация)': [log_reg_val_auc, xgb_val_auc]
})

# Форматируем таблицу
formatted_table = (models_results
                  .sort_values('ROC-AUC (валидация)', ascending=False)
                  .style
                  .format({'ROC-AUC (валидация)': '{:.4f}'})
                  .set_table_styles([{
                      'selector': 'th',
                      'props': [('font-weight', 'bold'),
                               ('text-align', 'center')]
                  }])
                  .hide(axis='index'))

print("\n=== СРАВНЕНИЕ МОДЕЛЕЙ ПО ROC-AUC ===")
print("="*45)
display(formatted_table)
print("="*45)

# Определяем лучшую модель
best_model_name = models_results.loc[models_results['ROC-AUC (валидация)'].idxmax(), 'Модель']
best_model_val_auc = models_results['ROC-AUC (валидация)'].max()

# Выводим итоговый результат
print(f"\nИТОГ: Лучшая модель - {best_model_name}")
print(f"Максимальный ROC-AUC на валидации: {best_model_val_auc:.4f}")

# Сравнительный анализ
improvement_pct = (xgb_val_auc - log_reg_val_auc) / log_reg_val_auc * 100

print(f"XGBoost превосходит логистическую регрессию на {abs(improvement_pct):.1f}%")
```

**Вывод:** Лучшая модель - XGBoost, её максимальный ROC-AUC на валидации - 0.89, что превосходит логистическую регрессию на 19.4%.
